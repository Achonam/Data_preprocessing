```python
isnnull = (df.isnull().mean()*100).round(2).sort_values(ascending=False)
isnnull
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_perct = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_report = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_perct.round(2)
})

missing_report
```

```python
df.columns
```

```python
for col in ["age","annual_income","monthly_spend"]:
    df[col]=df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

```python
for col in ['age', "annual_income", "monthly_spend"]:
    df[col] = df[col].fillna(df[col].median())
    
    df["contract_type"] = df["contract_type"].fillna("Unknown")
df.isnull().sum()
```

```python
df
```

```python
#Inspecting the duplicate rows
print ("Number of duplicates rows:", df.duplicated().sum())
df[df.duplicated(keep=False)].sort_values("customer_id")
```

```python
df = df.drop_duplicates()
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
#Inspecting inconsistent values
print(df["region"]. value_counts(dropna=False))
```

```python
df["region"] = (
    df["region"] 
    .astype("string")
    .str.strip()
    .str.title()
    
)
df["region"].value_counts(dropna=False)
```

```python
print("Invalid ages")
display(df[df["age"] < 0])

print("Invalid support-call count")
display(df[df["support_calls"] < 0])
```

```python
#converting invalid values to missing so they can be handled using 
# the same documented misssing_value strategy
df.loc[df["age"] < 0, "age"] = np.nan
df.loc[df["support_calls"] < 0, "support_calls"] = np.nan

df["age"] = df["age"].fillna(df["age"].median())
df["support_calls"] = df["support_calls"].fillna(df["support_calls"].median())

df[["age", "support_calls"]].describe()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.boxplot(df["annual_income"].dropna(), orientation="vertical")
ax.set_title("Annual Income - Boxplot")
ax.set_xlabel("Annual Income")
plt.show()
```

```python
df["annual_income"].describe(percentiles=[0.01, 0.25, 0.5, 0.75, 0.99])
```

```python
#investigating Outliers
Q1 = df["annual_income"].quantile(0.25)
print(Q1)
```

```python
Q3 = df["annual_income"].quantile(0.75)
print(Q3)
```

```python
IQR = Q3 - Q1 
print(IQR)
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower bound:",lower,
       
"Upper bound:", upper)
```

```python
outliers = df[
    (df["annual_income"] < lower) | 
    (df["annual_income"] > upper)
]

print("lower bound:", lower)
print("Upper bund:", upper)
display(outliers)
```

```python
print("Missing values:")
display(df.isnull().sum())

print("Duplicates rows:", df.duplicated().sum())

print("Negative ages:", (df["age"] < 0).sum())
print("Negative support calss:", (df["support_calls"] < 0).sum())

print("\nRegions:")
display(df["region"].value_counts())
```
